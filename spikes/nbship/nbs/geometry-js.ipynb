```javascript
//| export
/** Squared distance from the origin.
 * @param {number} x
 * @param {number} y
 * @returns {number}
 */
export function dist2(x, y) {
  return x * x + y * y;
}
```

```javascript
//| export
/** Greatest common divisor, by Euclid.
 * @param {number} a
 * @param {number} b
 * @returns {number}
 */
export function gcd(a, b) {
  return b === 0 ? a : gcd(b, a % b);
}
```

```javascript
//| export
/** A point, with the same methods the Rust notebook gives it. */
export class Point {
  constructor(x, y) { this.x = x; this.y = y; }
  dist2() { return dist2(this.x, this.y); }
  scaled(k) { return new Point(this.x * k, this.y * k); }
  toString() { return `Point(${this.x}, ${this.y})`; }
}
```

```javascript
const p = new Point(3, 4);
console.log(`${p} ${p.dist2()} ${p.scaled(2)}`);   // exploration, not exported
```
