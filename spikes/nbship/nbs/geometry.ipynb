# geometry

Explored in Leela, exported to a crate, shipped as a wheel.

```rust
//| default_exp geom
```

```rust
//| export
/// Squared distance from the origin.
pub fn dist2(x: i64, y: i64) -> i64 {
    x * x + y * y
}
```

```rust
//| export
/// Greatest common divisor, by Euclid.
pub fn gcd(a: u64, b: u64) -> u64 {
    if b == 0 { a } else { gcd(b, a % b) }
}
```

```rust
println!("{} {}", dist2(3, 4), gcd(48, 18));   // exploration, not exported
```

## A type, not just functions

The struct is plain Rust. `#[pyclass]` is written at export.

```rust
//| export
#[derive(Debug, Clone)]
pub struct Point {
    pub x: i64,
    pub y: i64,
}
```

```rust
//| export
impl Point {
    pub fn new(x: i64, y: i64) -> Self { Self { x, y } }
    pub fn dist2(&self) -> i64 { dist2(self.x, self.y) }
    pub fn scaled(&self, k: i64) -> Self { Self::new(self.x * k, self.y * k) }
}
```

```rust
let p = Point::new(3, 4);
println!("{:?} {} {:?}", p, p.dist2(), p.scaled(2));
```

## Fallible work, and what cannot cross

A `Result` becomes a Python exception. A generic function stays in the crate, and the export says why.

```rust
//| export
/// Parse a "x,y" pair into a point.
pub fn parse_point(s: String) -> Result<Point, std::num::ParseIntError> {
    let mut it = s.split(',');
    let x: i64 = it.next().unwrap_or("").trim().parse()?;
    let y: i64 = it.next().unwrap_or("").trim().parse()?;
    Ok(Point::new(x, y))
}
```

```rust
//| export
impl Point {
    /// The origin.
    pub fn origin() -> Self { Self::new(0, 0) }
    /// Move this point.
    pub fn nudge(&mut self, dx: i64, dy: i64) { self.x += dx; self.y += dy; }
}
```

```rust
//| export
/// Works in Rust, but PyO3 binds one concrete type at a time.
pub fn largest<T: PartialOrd>(v: Vec<T>) -> Option<T> {
    v.into_iter().reduce(|a, b| if a > b { a } else { b })
}
```

```rust
println!("{:?} {:?}", parse_point("3,4".to_string()), largest(vec![1, 9, 4]));
```
